```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/Dataset.csv")

# Keep a backup (important for reporting)
df_original = df.copy()

df.shape
```

```python
df_original = df.copy()
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

df.columns
```

```python
df.info()
```

```python
df = df.dropna(subset=["cuisines"])
```

```python
df.isnull().sum()
```

```python
binary_cols = [
    "has_online_delivery",
    "has_table_booking"
]

for col in binary_cols:
    df[col] = df[col].map({"Yes": 1, "No": 0})
```

```python
df["cuisines"] = df["cuisines"].str.strip()
```

```python
df = df.drop_duplicates()
```

```python
print("Original Shape:", df_original.shape)
print("Cleaned Shape:", df.shape)

rows_removed = df_original.shape[0] - df.shape[0]
percent_removed = round((rows_removed / df_original.shape[0]) * 100, 2)

print("Rows Removed:", rows_removed)
print("Percentage Removed:", percent_removed, "%")
```

```python
df.to_csv("../data/processed/cleaned_dataset.csv", index=False)
```
